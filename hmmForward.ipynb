```python
states = ["Rainy", "Sunny"]

# Observations
observations = ["Walk", "Shop", "Clean"]

# Observation sequence
sequence = ["Walk", "Shop", "Clean"]

# Initial state probabilities
initial_prob = {
    "Rainy": 0.4,
    "Sunny": 0.6
}

# State transition probabilities
transition_prob = {
    "Rainy": {"Rainy": 0.7, "Sunny": 0.3},
    "Sunny": {"Rainy": 0.4, "Sunny": 0.6}
}

# Emission probabilities
emission_prob = {
    "Rainy": {"Walk": 0.1, "Shop": 0.4, "Clean": 0.5},
    "Sunny": {"Walk": 0.6, "Shop": 0.3, "Clean": 0.1}
}
```

```python
# Initialize forward probabilities using the first observation
alpha = []
first_obs = sequence[0]
alpha_first = {}

for state in states:
    alpha_first[state] = initial_prob[state] * emission_prob[state][first_obs]
alpha.append(alpha_first)

# Calculate forward probabilities for remaining observations
for t in range(1, len(sequence)):
    obs = sequence[t]
    current = {}
    for current_state in states:
        total = 0.0
        for previous_state in states:
            total += (
                alpha[t - 1][previous_state]
                * transition_prob[previous_state][current_state]
            )
        current[current_state] = total * emission_prob[current_state][obs]
    alpha.append(current)

print("Forward probability table:")
for t, row in enumerate(alpha):
    print(f"t={t}, observation={sequence[t]}: {row}")

sequence_probability = sum(alpha[-1].values())
print("\nProbability of observation sequence:", sequence_probability)
```
