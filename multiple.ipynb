```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
X = np.array([
    [2, 60, 50],
    [3, 65, 55],
    [4, 70, 60],
    [5, 75, 65],
    [6, 80, 75],
    [7, 85, 80],
    [8, 90, 85],
    [9, 95, 90]
])
y = np.array([40, 45, 50, 60, 65, 75, 85, 90])
```

```python
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)
```

```python
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

```python
new_student = np.array([[6, 80, 75]])
predicted_score = model.predict(new_student)
print("Predicted Exam Score:", predicted_score[0])
```

```python
plt.scatter(y, y_pred, color='blue', label='Predicted Scores')
plt.plot([min(y), max(y)], [min(y), max(y)], color='green', label='Perfect Prediction')
plt.title('Multiple Linear Regression')
plt.xlabel('Actual Exam Score')
plt.ylabel('Predicted Exam Score')
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Multiple Regression R2 Score:", r2_score(y, y_pred))
```
