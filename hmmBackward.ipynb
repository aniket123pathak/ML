```python
states = ["Rainy", "Sunny"]

# Observations
observations = ["Walk", "Shop", "Clean"]

# Observation sequence
sequence = ["Walk", "Shop", "Clean"]

# Initial state probabilities
initial_prob = {
    "Rainy": 0.4,
    "Sunny": 0.6
}

# State transition probabilities
transition_prob = {
    "Rainy": {"Rainy": 0.7, "Sunny": 0.3},
    "Sunny": {"Rainy": 0.4, "Sunny": 0.6}
}

# Emission probabilities
emission_prob = {
    "Rainy": {"Walk": 0.1, "Shop": 0.4, "Clean": 0.5},
    "Sunny": {"Walk": 0.6, "Shop": 0.3, "Clean": 0.1}
}
```

```python
T = len(sequence)
beta = [{} for _ in range(T)]

# Base case: final backward probability is 1 for each state
for state in states:
    beta[T - 1][state] = 1.0

# Work backward through the observation sequence
for t in range(T - 2, -1, -1):
    next_obs = sequence[t + 1]
    for current_state in states:
        total = 0.0
        for next_state in states:
            total += (
                transition_prob[current_state][next_state]
                * emission_prob[next_state][next_obs]
                * beta[t + 1][next_state]
            )
        beta[t][current_state] = total

print("Backward probability table:")
for t, row in enumerate(beta):
    print(f"t={t}, observation={sequence[t]}: {row}")

sequence_probability = sum(
    initial_prob[state]
    * emission_prob[state][sequence[0]]
    * beta[0][state]
    for state in states
)
print("\nProbability of observation sequence:", sequence_probability)
```
