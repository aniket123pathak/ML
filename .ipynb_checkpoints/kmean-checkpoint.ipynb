# Iris Dataset — Clustering

This notebook loads the Iris dataset, checks the data, visualizes two features, standardizes all four features, then performs clustering and visualizes the resulting cluster labels.

**Note:** Cluster IDs such as `0`, `1`, and `2` are arbitrary labels; they do not automatically correspond to Iris species. The model does not use `y` during clustering.

## 1. Import libraries and load the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = iris.target

print("First five rows:")
print(X.head())
print("\nDataset shape:", X.shape)
print("\nMissing values:\n", X.isnull().sum())
print("\nTarget classes:", iris.target_names)
```

## 2. Visualize the original data

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=X,
    x="petal length (cm)",
    y="petal width (cm)"
)
plt.title("Iris Dataset Before Clustering")
plt.show()
```

## 3. Standardize the features

DBSCAN and distance-based clustering methods are sensitive to feature scales, so we standardize the four measurements before clustering.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 4. Create and fit the K-Means model

- `n_clusters=3`: ask the algorithm to find three clusters.
- `random_state=42`: makes the randomized initialization reproducible.
- `n_init=10`: tries ten initial centroid configurations and keeps the best result.

```python
from sklearn.cluster import KMeans

model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

labels = model.fit_predict(X_scaled)
print("Cluster labels:")
print(labels)
```

## 5. Visualize the clusters

The model clusters using all four standardized features. The graph displays petal length and petal width so the results can be viewed in 2D.

```python
X_plot = X[["petal length (cm)", "petal width (cm)"]].copy()
X_plot["Cluster"] = labels

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=X_plot,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="Cluster",
    palette="deep"
)
plt.title("K-Means Clustering on Iris Dataset")
plt.show()
```

## Quick notes

- K-Means assigns points to the nearest centroid and updates centroids using the mean of their assigned points.
- You must specify `n_clusters` in advance.
- Cluster IDs are not species names; `y` is kept only for reference and is not passed to the model.
