# Polynomial Regression — Iris Dataset

Predict **petal width** from **petal length** using polynomial features. Preprocessing and Iris visualization follow the supplied AdaBoost notebook.

## 1. Iris data preprocessing (reference format)

This preprocessing follows the supplied AdaBoost notebook: load Iris, create `X` as a pandas DataFrame, keep `y = iris.target` for the species labels, inspect the data, and make the same petal-length vs petal-width Seaborn scatterplot colored by species.

For the regression notebooks, the original class labels are used for this dataset inspection only. The regression task then selects a **continuous Iris measurement** as its numeric target, which is required for regression.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load Dataset — same as the reference notebook
iris = load_iris()
```

```python
# Create DataFrame X and class-label y — same as the reference
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = iris.target
```

```python
# Inspect the data — same as the reference
print(X.head())
print(X.shape)
print(X.isnull().sum())
print(iris.target_names)
```

```python
# Visualization — same features, style, and species hue as the reference
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=X,
    x="petal length (cm)",
    y="petal width (cm)",
    hue=iris.target_names[y]
)
plt.title("Iris Dataset Visualization")
plt.show()
```

## 2. Select regression feature and split

Polynomial regression first transforms the single input into powers such as \(x\), \(x^2\), and so on, then fits a linear regression model to those transformed features.

```python
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

X_reg = X[["petal length (cm)"]]
y_reg = X["petal width (cm)"]

X_train, X_test, y_train, y_test = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

## 3. Train Polynomial Regression (degree = 2)

```python
degree = 2

model = Pipeline([
    ("polynomial_features", PolynomialFeatures(degree=degree, include_bias=False)),
    ("linear_regression", LinearRegression())
])

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

linear_model = model.named_steps["linear_regression"]
print("Polynomial degree:", degree)
print("Intercept:", round(linear_model.intercept_, 4))
print("Coefficients:", linear_model.coef_)
```

## 4. All regression scores

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE:  {mae:.4f}")
print(f"MSE:  {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²:   {r2:.4f}")
```

```python
# Plot observed data and polynomial curve
x_line = pd.DataFrame({
    "petal length (cm)": np.linspace(
        X_reg["petal length (cm)"].min(),
        X_reg["petal length (cm)"].max(),
        300
    )
})
y_line = model.predict(x_line)

plt.figure(figsize=(8, 5))
plt.scatter(X_reg["petal length (cm)"], y_reg, alpha=0.6, label="Observed data")
order = np.argsort(x_line["petal length (cm)"].to_numpy())
plt.plot(x_line.iloc[order, 0], y_line[order], label=f"Polynomial curve (degree {degree})")
plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.title("Polynomial Regression on Iris")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Effect of different polynomial degrees

Compare degrees 1 through 5. Higher degree can fit more curved relationships, but very high-degree models may overfit.

```python
degree_results = []

for d in range(1, 6):
    degree_model = Pipeline([
        ("polynomial_features", PolynomialFeatures(degree=d, include_bias=False)),
        ("linear_regression", LinearRegression())
    ])
    degree_model.fit(X_train, y_train)
    train_pred = degree_model.predict(X_train)
    test_pred = degree_model.predict(X_test)

    degree_results.append({
        "Degree": d,
        "Train R²": r2_score(y_train, train_pred),
        "Test R²": r2_score(y_test, test_pred),
        "Test MAE": mean_absolute_error(y_test, test_pred),
        "Test RMSE": np.sqrt(mean_squared_error(y_test, test_pred))
    })

degree_results_df = pd.DataFrame(degree_results)
display(degree_results_df.round(4))
```

```python
plt.figure(figsize=(8, 5))
plt.plot(degree_results_df["Degree"], degree_results_df["Train R²"], marker="o", label="Train R²")
plt.plot(degree_results_df["Degree"], degree_results_df["Test R²"], marker="s", label="Test R²")
plt.xlabel("Polynomial degree")
plt.ylabel("R²")
plt.title("Effect of Polynomial Degree")
plt.xticks(range(1, 6))
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## Viva points

- Polynomial regression models curved relationships by adding powers of the input features.
- `PolynomialFeatures` generates terms such as \(x^2\) and \(x^3\).
- Increasing degree increases flexibility but can cause overfitting.
- Use regression metrics to compare the model's predictions with actual values.
