# Iris Dataset — Clustering

This notebook loads the Iris dataset, checks the data, visualizes two features, standardizes all four features, then performs clustering and visualizes the resulting cluster labels.

**Note:** Cluster IDs such as `0`, `1`, and `2` are arbitrary labels; they do not automatically correspond to Iris species. The model does not use `y` during clustering.

**Extra dependency:** K-Medoids is provided here by `scikit-learn-extra`, not the standard `sklearn.cluster` module.

## 0. Install the K-Medoids package

## 1. Import libraries and load the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = iris.target

print("First five rows:")
print(X.head())
print("\nDataset shape:", X.shape)
print("\nMissing values:\n", X.isnull().sum())
print("\nTarget classes:", iris.target_names)
```

## 2. Visualize the original data

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=X,
    x="petal length (cm)",
    y="petal width (cm)"
)
plt.title("Iris Dataset Before Clustering")
plt.show()
```

## 3. Standardize the features

DBSCAN and distance-based clustering methods are sensitive to feature scales, so we standardize the four measurements before clustering.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 4. Create and fit the K-Medoids model

- `n_clusters=3`: ask the algorithm to find three clusters.
- `random_state=42`: makes randomized initialization reproducible.
- `metric="euclidean"`: uses Euclidean distance.

A **medoid** is an actual data point selected to represent a cluster, unlike a K-Means centroid, which is the mean position of its assigned points.

```python
from sklearn_extra.cluster import KMedoids

model = KMedoids(
    n_clusters=3,
    random_state=42,
    metric="euclidean"
)

labels = model.fit_predict(X_scaled)
print("Cluster labels:")
print(labels)
```

## 5. Visualize the clusters

The model clusters using all four standardized features. The graph shows two original features. Black X markers indicate the medoids.

```python
X_plot = X[["petal length (cm)", "petal width (cm)"]].copy()
X_plot["Cluster"] = labels

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=X_plot,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="Cluster",
    palette="deep"
)

medoid_indices = model.medoid_indices_
medoids = X.iloc[medoid_indices]
plt.scatter(
    medoids["petal length (cm)"],
    medoids["petal width (cm)"],
    marker="X",
    s=180,
    color="black",
    label="Medoids"
)

plt.title("K-Medoids Clustering on Iris Dataset")
plt.legend()
plt.show()
```

## Quick notes

- K-Medoids uses actual samples as cluster representatives.
- `fit_predict()` fits the model and returns a cluster ID for each sample.
- `model.medoid_indices_` contains the row indices of the selected medoids.
- Cluster IDs are not species names; `y` is not used for clustering.
