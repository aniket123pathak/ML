# Simple Linear Regression — Iris Dataset

Predict **petal width** using only **petal length**. The preprocessing/inspection cells match the supplied AdaBoost notebook. The remaining cells train the built-in `LinearRegression` model, evaluate it, and visualize its fitted line.

## 1. Iris data preprocessing (reference format)

This preprocessing follows the supplied AdaBoost notebook: load Iris, create `X` as a pandas DataFrame, keep `y = iris.target` for the species labels, inspect the data, and make the same petal-length vs petal-width Seaborn scatterplot colored by species.

For the regression notebooks, the original class labels are used for this dataset inspection only. The regression task then selects a **continuous Iris measurement** as its numeric target, which is required for regression.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

# Load Dataset — same as the reference notebook
iris = load_iris()
```

```python
# Create DataFrame X and class-label y — same as the reference
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = iris.target
```

```python
# Inspect the data — same as the reference
print(X.head())
print(X.shape)
print(X.isnull().sum())
print(iris.target_names)
```

```python
# Visualization — same features, style, and species hue as the reference
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=X,
    x="petal length (cm)",
    y="petal width (cm)",
    hue=iris.target_names[y]
)
plt.title("Iris Dataset Visualization")
plt.show()
```

## 2. Select regression features and split the data

- Input (`X_reg`): petal length only.
- Target (`y_reg`): petal width.
- The target is continuous, so this is a regression task.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

X_reg = X[["petal length (cm)"]]
y_reg = X["petal width (cm)"]

X_train, X_test, y_train, y_test = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

## 3. Train the Linear Regression model

```python
model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])
print(f"Regression equation: petal width = {model.intercept_:.4f} + ({model.coef_[0]:.4f} × petal length)")
```

## 4. All regression scores

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print(f"MAE:  {mae:.4f}")
print(f"MSE:  {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²:   {r2:.4f}")
```

```python
# Actual vs predicted values
plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred, alpha=0.8)
low = min(float(y_test.min()), float(np.min(y_pred)))
high = max(float(y_test.max()), float(np.max(y_pred)))
plt.plot([low, high], [low, high], linestyle="--")
plt.xlabel("Actual petal width (cm)")
plt.ylabel("Predicted petal width (cm)")
plt.title("Linear Regression: Actual vs Predicted")
plt.grid(alpha=0.3)
plt.show()
```

```python
# Plot data points and the fitted regression line
x_line = pd.DataFrame({
    "petal length (cm)": np.linspace(
        X_reg["petal length (cm)"].min(),
        X_reg["petal length (cm)"].max(),
        200
    )
})
y_line = model.predict(x_line)

plt.figure(figsize=(8, 5))
plt.scatter(X_reg["petal length (cm)"], y_reg, alpha=0.65, label="Observed data")
order = np.argsort(x_line["petal length (cm)"].to_numpy())
plt.plot(x_line.iloc[order, 0], y_line[order], label="Regression line")
plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.title("Simple Linear Regression on Iris")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## Viva points

- Simple linear regression uses one independent variable to predict one continuous target.
- The coefficient is the expected change in the predicted target for a one-unit increase in the input.
- MAE, MSE, and RMSE measure prediction error; lower is better.
- R² measures how much target variation is explained relative to predicting the mean; it can be negative on a test set.
