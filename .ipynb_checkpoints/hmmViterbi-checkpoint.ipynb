# HMM Viterbi Algorithm

Find the most likely hidden-state sequence for the observation sequence.

## 1. Input probabilities
Run this cell first. It uses the exact states, observations, sequence, and probabilities provided.

```python
states = ["Rainy", "Sunny"]

# Observations
observations = ["Walk", "Shop", "Clean"]

# Observation sequence
sequence = ["Walk", "Shop", "Clean"]

# Initial state probabilities
initial_prob = {
    "Rainy": 0.4,
    "Sunny": 0.6
}

# State transition probabilities
transition_prob = {
    "Rainy": {"Rainy": 0.7, "Sunny": 0.3},
    "Sunny": {"Rainy": 0.4, "Sunny": 0.6}
}

# Emission probabilities
emission_prob = {
    "Rainy": {"Walk": 0.1, "Shop": 0.4, "Clean": 0.5},
    "Sunny": {"Walk": 0.6, "Shop": 0.3, "Clean": 0.1}
}
```

## Viterbi algorithm
Viterbi finds the single most likely hidden-state path. It uses `max` instead of summing over previous states. A backpointer remembers the best previous state.

```python
# Viterbi score table and backpointer table
viterbi = []
backpointer = []

# Initialize using the first observation
first_obs = sequence[0]
first_scores = {}
first_backpointers = {}

for state in states:
    first_scores[state] = initial_prob[state] * emission_prob[state][first_obs]
    first_backpointers[state] = None

viterbi.append(first_scores)
backpointer.append(first_backpointers)

# Find the best path score for each later observation
for t in range(1, len(sequence)):
    obs = sequence[t]
    current_scores = {}
    current_backpointers = {}

    for current_state in states:
        candidates = {
            previous_state:
                viterbi[t - 1][previous_state]
                * transition_prob[previous_state][current_state]
                * emission_prob[current_state][obs]
            for previous_state in states
        }

        best_previous_state = max(candidates, key=candidates.get)
        current_scores[current_state] = candidates[best_previous_state]
        current_backpointers[current_state] = best_previous_state

    viterbi.append(current_scores)
    backpointer.append(current_backpointers)

print("Viterbi score table:")
for t, row in enumerate(viterbi):
    print(f"t={t}, observation={sequence[t]}: {row}")

# Backtrack from the best final state
last_state = max(viterbi[-1], key=viterbi[-1].get)
best_path = [last_state]

for t in range(len(sequence) - 1, 0, -1):
    best_path.append(backpointer[t][best_path[-1]])

best_path.reverse()
print("\nMost likely hidden-state sequence:", best_path)
print("Probability of best path:", max(viterbi[-1].values()))
```

## Result
The output path is the most likely hidden-state sequence for the observations. The final score is the joint probability of that path and the observations.
